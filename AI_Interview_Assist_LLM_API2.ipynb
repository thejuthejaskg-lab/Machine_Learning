```python
from groq import Groq
from google.colab import userdata
key = userdata.get('GROQ_API_KEY')
client = Groq(api_key=key)
print("Groq API connected sucessfully")
```

```python
models = client.models.list()
print ("Available GROQ Models:\n")
for model in models.data:
  print(model.id)
```

```python
from groq import Groq
response_1 = client.chat.completions.create(
    model = "qwen/qwen3.8-27b",
    messages = [
        {
          "role":"system",
          "content": "BUild an machine learning expert app."
        },
        {
          "role" : "user",
          "content" :"""
          Explain CNN 5 points and explain
          architecture of this module in 5 points
          covering each points with explinations
          """
        }
    ],
    temperature=0.7,
    max_tokens=1000
)
print(response_1.choices[0].message.content)
```
