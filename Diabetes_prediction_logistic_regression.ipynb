```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,f1_score, confusion_matrix, precision_score, recall_score
```

```python
# 2. Load dataset file_path = "/content/CAR DETAILS FROM CAR DEKHO.excel"
df = pd.read_excel("/content/diabetesxl.xlsx")
print("--- Dataset Head ---")
print(df.head(10))
```

```python
# 3. Find null values and drop
print("--- Null Values Before Drop ---")
print(df.isnull().sum())
df = df.dropna()
print("\n--- Null Values After Drop ---")
print(df.isnull().sum())
print()
```

```python
# 4. Check duplicates and drop
print(f"Duplicates before: {df.duplicated().sum()}")
df = df.drop_duplicates()
print(f"Duplicates after:  {df.duplicated().sum()}")
print()
```

```python
x = df[['age','mass','insu','plas']]
y = df['class']
x = df[['age' , 'mass' , 'insu' , 'plas']]
y = df['class']
print("--- Features X (head) ---")
print(x.head(10))
print("\n--- Target Y (head) ---")
print(y.head(10))
print()
```

```python
# 7. Split Train Test (20%)
xtrain, xtest, ytrain, ytest = train_test_split( x, y, test_size=0.2, random_state=42 )
print(f"Xtrain shape: {xtrain.shape}")
print(f"Xtest shape:  {xtest.shape}")
print()
```

```python
Logr = LogisticRegression()
```

```python

 Logr.fit(xtrain, ytrain)
```

```python
y_pred = Logr.predict(xtest)
result = pd.DataFrame({
    'Actual': ytest,
    'Predicted': y_pred
})
print(result)
```

```python
age = int(input("enter the age:"))
mass = int(input("enter the mass:"))
insulin = int(input("enter the insulin level:"))
plasma = int(input("enter the plasma level:"))
pred = Logr.predict([[age,mass,insulin,plasma]])
for tested_positive in pred:
  if pred == tested_positive:
    print("plese consult the doctor")
  else:
    print("eat more laddu stay happy")
```

```python
print("accuracy:",accuracy_score(ytest,y_pred))
print("precision:",precision_score(ytest,y_pred,pos_label="tested_positive"))
print("recall:",recall_score(ytest,y_pred,pos_label="tested_positive"))
print("f1 score:",f1_score(ytest,y_pred,pos_label="tested_positive"))
print("confusion matrix:\n",confusion_matrix(ytest,y_pred))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
cm = confusion_matrix(ytest,y_pred)
sns.heatmap(cm,annot=True)
plt.xlabel("predicicted")
plt.ylabel("Actual")
plt.title("confusion matrix")
plt.show()
```
