```python
from groq import Groq
from google.colab import userdata
key = userdata.get('GROQ_API_KEY')
client = Groq(api_key=key)
print("Groq API connected sucessfully")
```

```python
models = client.models.list()
print ("Available GROQ Models:\n")
for model in models.data:
  print(model.id)
```

```python
from groq import Groq
response = client.chat.completions.create(
    model = "openai/gpt-oss-120b",
    messages = [
        {
          "role":"system",
          "content": "you are an experienced technical interview."

        },
        {
          "role" : "user",
          "content" :"""
          Generate 5 intermediate - level python
          interview question for final - year
          engineering students.
          include conceptual and particical question.
          """
        }
    ],
    temperature=0.7,
    max_tokens=1000
)
print(response.choices[0].message.content)
```
