```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,precision_score,recall_score, confusion_matrix, classification_report,f1_score
```

```python
df = pd.read_excel("/content/diabetesxl.xlsx")
print("--- Dataset Head ---")
print(df.head(10))
```

```python
print("--- Null Values Before Drop ---")
print(df.isnull().sum())
df = df.dropna()
print("\n--- Null Values After Drop ---")
print(df.isnull().sum())
print()
```

```python
print(f"Duplicates before: {df.duplicated().sum()}")
df = df.drop_duplicates()
print(f"Duplicates after:  {df.duplicated().sum()}")
print()
```

```python
x = df[['age','mass','insu','plas']]
y = df['class']
x = df[['age' , 'mass' , 'insu' , 'plas']]
y = df['class']
print("--- Features X (head) ---")
print(x.head(10))
print("\n--- Target Y (head) ---")
print(y.head(10))
print()
```

```python
xtrain, xtest, ytrain, ytest = train_test_split( x, y, test_size=0.2, random_state=42 )
print(f"Xtrain shape: {xtrain.shape}")
print(f"Xtest shape:  {xtest.shape}")
print()
```

```python
random_decision = RandomForestClassifier(n_estimators=100 , random_state=42 )
random_decision.fit(xtrain,ytrain)
```

```python
y_pred = random_decision.predict(xtest)
result = pd.DataFrame({
    'Actual': ytest,
    'Predicted': y_pred
})
print(result)
```

```python
age = int(input("enter the age:"))
mass = int(input("enter the mass:"))
insulin = int(input("enter the insulin level:"))
plasma = int(input("enter the plasma level:"))
pred = random_decision.predict([[age,mass,insulin,plasma]])
print(pred)
```

```python
import matplotlib.pyplot as plt
cm = confusion_matrix(ytest,y_pred)
sns.heatmap(cm,annot=True)
plt.xlabel("predicicted")
plt.ylabel("Actual")
plt.title(" random confusion matrix")
plt.show()
```
