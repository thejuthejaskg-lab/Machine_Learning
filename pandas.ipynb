```python
df.isnull().sum()
```

```python
df.dropna(inplace = True) #inplace -> save the changes in same df
df.isnull().sum()
```

```python
print("number of dulicated rows=",df.duplicated().sum())
```

```python
df.drop_duplicates(inplace = True)
print("number of dulicated rows=",df.duplicated().sum())
```

create colums for total and average marks

```python
#df["new_col"]=expression
df["Total"] = (df[subjects_cols]).sum(axis = 1)
df.head()
```

```python
#df["new_col"]=expression
df["Average"] = (df[subjects_cols]).mean(axis = 1)
df.head()
```

```python
overall_avg = df["Average"].mean()
print("Overall Average of Student Averages:", overall_avg)
```

```python
overall_avg = df["Average"].mean()
print("Overall Average of Student Averages:", round(overall_avg,4))
```
